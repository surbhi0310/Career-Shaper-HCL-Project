```python

# Function to take inputs from the user
def get_inputs():

    while True:
        try:
            # Ask the user for three inputs
            units_input = input("\nEnter units sold (or 'end' to stop): ")
            if units_input.lower() == "end":
                return None

            price_input = input("Enter price per unit (or 'end' to stop): ")
            if price_input.lower() == "end":
                return None

            target_input = input("Enter sales target (or 'end' to stop): ")
            if target_input.lower() == "end":
                return None

            # Convert inputs into required data types
            units = int(units_input)
            price = float(price_input)
            target = float(target_input)

            return units, price, target

        except ValueError:
            # Handle invalid numeric input
            print("Invalid input! Please enter valid numbers.")
            print("Please try again.")


# Function to calculate sales performance
def sales_performance(units, price, target):

    # Calculate total revenue
    revenue = units * price

    # Calculate target achievement
    achievement = (revenue / target) * 100

    # Display results
    print("\n----- Sales Performance -----")
    print(f"Total Revenue: Rs. {revenue:.2f}")
    print(f"Target Achievement: {achievement:.2f}%")
    print("-----------------------------")


# Main program using a while loop
while True:

    # Get input from the user
    data = get_inputs()

    # Stop the program if user enters 'end'
    if data is None:
        print("\nProgram stopped. Thank you!")
        break

    try:
        # Calculate and display sales performance
        sales_performance(*data)

    except ZeroDivisionError:
        # Handle zero sales target
        print("Error: Sales target cannot be zero!")
        print("Please enter the details again.")
        continue
```
